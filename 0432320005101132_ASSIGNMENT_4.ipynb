```python
import numpy as np
from scipy import stats

# Given data
data1 = np.array([18.4, 19.1, 17.9, 18.7, 18.3])
data2 = np.array([20.2, 20.5, 20.1, 20.3, 20.4])

# H0: μ1 = μ2
# H1: μ1 ≠ μ2

# Sample statistics
mean1 = np.mean(data1)
mean2 = np.mean(data2)

var1 = np.var(data1, ddof=1)
var2 = np.var(data2, ddof=1)

mean_diff = mean1 - mean2

print("Sample Mean 1:", mean1)
print("Sample Mean 2:", mean2)
print("Sample Variance 1:", var1)
print("Sample Variance 2:", var2)
print("Mean Difference:", mean_diff)

# Welch's Two-Sample t-test
t_stat, p_value = stats.ttest_ind(data1, data2, equal_var=False)

# Degrees of freedom (Welch-Satterthwaite equation)
n1 = len(data1)
n2 = len(data2)

df = ((var1/n1 + var2/n2)**2) / (
    ((var1/n1)**2)/(n1-1) +
    ((var2/n2)**2)/(n2-1)
)

print("\nT-statistic:", t_stat)
print("Degrees of Freedom:", df)
print("Two-tailed P-value:", p_value)

# 95% Confidence Interval
alpha = 0.05

se = np.sqrt(var1/n1 + var2/n2)

t_critical = stats.t.ppf(1 - alpha/2, df)

margin_error = t_critical * se

ci_lower = mean_diff - margin_error
ci_upper = mean_diff + margin_error

print("\n95% Confidence Interval:")
print(f"({ci_lower:.4f}, {ci_upper:.4f})")

# Decision
if p_value < alpha:
    print("\nDecision: Reject H0")
    print("Interpretation: There is a statistically significant difference between the population means.")
else:
    print("\nDecision: Fail to Reject H0")
    print("Interpretation: There is not enough evidence to conclude that the population means differ.")
```
